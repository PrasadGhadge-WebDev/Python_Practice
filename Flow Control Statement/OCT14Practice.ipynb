```python

employees = {
    101: {"name": "Amit", "department": "IT", "salary": 45000},
    102: {"name": "Priya", "department": "HR", "salary": 38000},
    103: {"name": "Sneha", "department": "IT", "salary": 52000},
    104: {"name": "Raj", "department": "Sales", "salary": 41000},
    105: {"name": "Anjali", "department": "HR", "salary": 47000}
}

# 6. Print employee ID with employee name
print("\n6. Employee ID and Name")

for emp_id, data in employees.items():
    print(emp_id, data["name"])


# 7. Print employees who belong to the IT department
print("\n7. IT Department Employees")

for emp_id, data in employees.items():
    if data["department"] == "IT":
        print(emp_id, data["name"])


# 8. Print employees whose salary is greater than 40,000
print("\n8. Employees with Salary Greater Than 40000")

for emp_id, data in employees.items():
    if data["salary"] > 40000:
        print(emp_id, data["name"], data["salary"])


# 9. Find the employee with the highest salary
print("\n9. Employee with Highest Salary")

max_id = None
max_salary = -1

for emp_id, data in employees.items():
    if data["salary"] > max_salary:
        max_salary = data["salary"]
        max_id = emp_id

print(max_id, employees[max_id]["name"], max_salary)


# 10. Calculate the average salary of all employees
print("\n10. Average Salary")

total_salary = 0

for emp_id, data in employees.items():
    total_salary += data["salary"]

average_salary = total_salary / len(employees)

print("Average Salary:", average_salary)
```

```python
products = {
    "P101": {"name": "Laptop", "price": 55000, "stock": 8},
    "P102": {"name": "Mouse", "price": 700, "stock": 25},
    "P103": {"name": "Keyboard", "price": 1200, "stock": 15},
    "P104": {"name": "Monitor", "price": 14000, "stock": 5},
    "P105": {"name": "Headphones", "price": 2500, "stock": 0}
}


# 11. Print the name and price of every product
print("\n11. Product Name and Price")

for product_id, data in products.items():
    print(data["name"], data["price"])


# 12. Print products having stock less than 10
print("\n12. Products with Stock Less Than 10")

for product_id, data in products.items():
    if data["stock"] < 10:
        print(product_id, data["name"], data["stock"])


# 13. Print products that are out of stock
print("\n13. Out of Stock Products")

for product_id, data in products.items():
    if data["stock"] == 0:
        print(product_id, data["name"])


# 14. Calculate the total value of each product using price * stock
print("\n14. Total Value of Each Product")

for product_id, data in products.items():
    total_value = data["price"] * data["stock"]
    print(data["name"], total_value)


# 15. Apply a 5% discount to products costing more than 10,000
print("\n15. Products After 5% Discount")

for product_id, data in products.items():
    if data["price"] > 10000:
        discount = data["price"] * 0.05
        final_price = data["price"] - discount

        print(
            data["name"],
            "Original Price:", data["price"],
            "Discount:", discount,
            "Final Price:", final_price
        )
```

```python
# 16. Merge two dictionaries without using the | operator
print("\n16. Merge Two Dictionaries")

dict1 = {"a": 10, "b": 20}
dict2 = {"c": 30, "d": 40}

merged_dict = dict1.copy()
merged_dict.update(dict2)

print(merged_dict)


# 17. Find common keys between two dictionaries
print("\n17. Common Keys")

dict1 = {"a": 10, "b": 20, "c": 30}
dict2 = {"b": 200, "c": 300, "d": 400}

common_keys = []

for key in dict1:
    if key in dict2:
        common_keys.append(key)

print(common_keys)


# 18. Create a dictionary from two separate lists
print("\n18. Dictionary from Two Lists")

keys = ["name", "age", "city"]
values = ["Rahul", 22, "Pune"]

new_dict = dict(zip(keys, values))

print(new_dict)


# 19. Count the frequency of every character in a user-given string
print("\n19. Character Frequency")

text = input("Enter a string: ")

frequency = {}

for char in text:
    if char in frequency:
        frequency[char] += 1
    else:
        frequency[char] = 1

print(frequency)


# 20. Remove duplicate values while keeping the first matching key
print("\n20. Remove Duplicate Values")

original_dict = {
    "a": 10,
    "b": 20,
    "c": 10,
    "d": 30,
    "e": 20
}

unique_dict = {}

for key, value in original_dict.items():
    if value not in unique_dict.values():
        unique_dict[key] = value

print(unique_dict)
```
