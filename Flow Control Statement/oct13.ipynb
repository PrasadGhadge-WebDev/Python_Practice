```python
input = [[10,20,30] , [40,50] , [60,70,80]]
flatten_list = []

for data in input:
    for ele in data:
        flatten_list. append(ele)

print(flatten_list)
```

```python
input = [[10,21,30] , [41,50] , [60,71,81]]
odd_ele_flatten_list = []

for data in input:
    for ele in data:
        if ele % 2 != 0:
            odd_ele_flatten_list.append(ele)

print(odd_ele_flatten_list)
```

```python
input = [90,[10,20,30],[40,50] ,77, [60,70,80],55]
flatten_list = []
for data in input:
    if type(data) == list:
        for ele in data:
            flatten_list.append(ele)
    else:
        flatten_list.append(data)
print(flatten_list)
```

```python
input = ["90", [10,20,30], (40,50) , "77", {60,70,80}, 55]
flatten_list = []
for data in input:
    if isinstance(data ,(list,tuple,set)):
        flatten_list.extend(data)

    elif isinstance(data,str):
        flatten_list.append(int(data))
    else:
        flatten_list.append(data)
print(flatten_list)
```
