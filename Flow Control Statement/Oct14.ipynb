```python
marks = [70, 80, 65, 90]
print(marks)
marks.append(95)
print(marks)
marks[0] = 75
print(marks)
marks.insert(2, 99)
print(marks)
marks.extend([60, 70])
print(marks)
marks.remove(99)
print(marks)
marks.pop(3)
print(marks)
marks.sort()
print(marks)
print(len(marks))
print(max(marks))
print(min(marks))
print(sum(marks))
```

```python
student = {
    "name": "Rahul",
    "age": 22,
    "course": "Python",
    "city": "Pune",
    "marks": 78
}

# 1. Print only the student's name
print("Student Name:", student["name"])

# 2. Display all dictionary keys
print("Dictionary Keys:", student.keys())

# 3. Display all key-value pairs
print("Key-Value Pairs:", student.items())

# 4. Add a new key "email"
student["email"] = "rahul@gmail.com"
print("After Adding Email:", student)

# 5. Use get() to display the phone key without causing an error
print("Phone:", student.get("phone"))
```

```python

employees = {
    101: {"name": "Amit", "department": "IT", "salary": 45000},
    102: {"name": "Priya", "department": "HR", "salary": 38000},
    103: {"name": "Sneha", "department": "IT", "salary": 52000},
    104: {"name": "Raj", "department": "Sales", "salary": 41000},
    105: {"name": "Anjali", "department": "HR", "salary": 47000}
}

# 6. Print employee ID with employee name
print("6. Employee ID and Name:")
for emp_id, details in employees.items():
    print(emp_id, details["name"])

# 7. Print employees who belong to the IT department
print("\n7. IT Department Employees:")
for emp_id, details in employees.items():
    if details["department"] == "IT":
        print(emp_id, details["name"])

# 8. Print employees whose salary is greater than 40,000
print("\n8. Employees with Salary Greater Than 40000:")
for emp_id, details in employees.items():
    if details["salary"] > 40000:
        print(emp_id, details["name"], details["salary"])

# 9. Find the employee with the highest salary
print("\n9. Employee with Highest Salary:")
highest_employee = max(
    employees.items(),
    key=lambda item: item[1]["salary"]
)
print("Employee ID:", highest_employee[0])
print("Name:", highest_employee[1]["name"])
print("Salary:", highest_employee[1]["salary"])

# 10. Calculate the average salary of all employees
print("\n10. Average Salary:")
total_salary = 0

for details in employees.values():
    total_salary += details["salary"]

average_salary = total_salary / len(employees)
print("Average Salary:", average_salary)
```
