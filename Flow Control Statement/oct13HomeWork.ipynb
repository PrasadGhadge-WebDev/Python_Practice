### Remove duplicates while keeping the original order

```python
l = [5, 4, 5, 4, 6, 7, 8, 4, 3, 5, 3, 4]
unique_list = list(dict.fromkeys(l))
print(unique_list)
```

### Count the frequency of each number

```python
l = [5, 4, 5, 4, 6, 7, 8, 4, 3, 5, 3, 4]
freq_ele = {}
for item in l:
    freq_ele[item] = freq_ele.get(item, 0) + 1
print(freq_ele)
```

### Count the frequency of each character, including spaces

```python
s = "I love python programming"
freq_char = {}
for char in s:
    freq_char[char] = freq_char.get(char, 0) + 1
print(freq_char)
```
